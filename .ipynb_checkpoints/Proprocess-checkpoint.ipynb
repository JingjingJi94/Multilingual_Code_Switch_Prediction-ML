```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
import torch
from preprocess_util import preprocess_and_label
from transformers import AutoTokenizer
from huggingface_hub import login
from datasets import load_dataset


# prompt to enter your Hugging Face API token for acessing HF account
login()
```

```python
# Load dataset directly from Hugging Face
dataset = load_dataset("Shelton1013/SwitchLingua_text")

# See available splits
print(dataset)

# Convert train split to pandas
df = dataset["train"].to_pandas()

print("Dataset shape:", df.shape)
print(df.head())
```

```python
# Language Pairs

# Create language pair column
df["language_pair"] = df["first_language"] + "-" + df["second_language"]

# Number of unique pairs
print("Unique pairs:", df["language_pair"].nunique())

# Count the  number of samples per language pair
pair_counts = df["language_pair"].value_counts()
print("Language pair counts:\n", pair_counts)

plt.figure(figsize=(6,4))
sns.barplot(
    x=pair_counts.index,
    y=pair_counts.values
)

plt.xticks(rotation=45)
plt.title("Distribution of Language Pairs in SwitchLingua")
plt.ylabel("Number of Samples")
plt.xlabel("Language Pair")
plt.tight_layout()
plt.show()
```

```python

chosen_pairs = [("Chinese", "English"), ("Hindi", "English"), ("Spanish", "English")]

# df_filtered only contains samples with the chosen language pairs.
df_filtered = df[df["language_pair"].isin([f"{l1}-{l2}" for l1,l2 in chosen_pairs])].reset_index(drop=True)

print(f"Filtered dataset shape: {df_filtered.shape}")
print(df_filtered.head())
print(df_filtered.loc[0, "data_generation_result"])
```

```python
# Create tokenizer instance
main_model_name = "xlm-roberta-base"
tokenizer = AutoTokenizer.from_pretrained(main_model_name)
```

```python
# this section help visualize data on 3 test samples, one from each language pair selected.
# The next section treats entire dataset for selected language pair

# Filter df_filtered for only those pairs
df_subset = df_filtered[df_filtered["language_pair"].isin(
    [f"{l1}-{l2}" for l1, l2 in chosen_pairs]
)].copy()

# Take only the first row for each pair
df_subset = df_subset.groupby("language_pair").head(1).reset_index(drop=True)

# Step 2: preprocess and label
df_subset = preprocess_and_label(df_subset, tokenizer)

# Step 3: display preprocessed column as tables
for idx, row in df_subset.iterrows():
    pre = row["preprocessed"]
    print(f"Row {idx} | Language Pair: {row['language_pair']}")
    print("Original Text:")
    print(pre["original_text"])
    print("\nToken-level table:")

    table = pd.DataFrame({
        "token": pre["tokens"],
        "lang_id": pre["lang_ids"],
        "ysw": pre["ysw"],
        "ydur": pre["ydur"]
    })

    # show first 30 tokens for readability, remove .head() to show all
    display(table.head(40))
    
    print("\n" + "="*80 + "\n")
```

```python
# Apply tokenization and labelling to entire dataset with chosen language pair
print("Starting preprocessing of entire dataset...")
df_filtered = preprocess_and_label(df_filtered, tokenizer)

#Save the preprocessed DataFrame to a pickle
output_path = "./df_preprocessed.pkl"
with open(output_path, "wb") as f:
    pickle.dump(df_filtered, f)
print(f"Preprocessed dataset saved to {output_path}")
```
